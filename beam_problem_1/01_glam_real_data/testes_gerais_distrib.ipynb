```python
import numpy as np
n_latent_samples = 2000
cov_mean   = 1.0
cov_cov    = 0.02
sigma_cov  = np.sqrt(np.log(1 + cov_cov**2))
mu_cov     = np.log(cov_mean) - sigma_cov**2 / 2
cov_latent = np.random.lognormal(mean=mu_cov, sigma=sigma_cov, size=n_latent_samples)
cov_latent
```

```python
# CRIAR BANCO DE DADOS COM TODOS OS DATASETS
import os, dill, pandas as pd, sqlite3

diretorio = '/home/tempestade/Documentos/2024-1_victor_hugo_renata_maria/beam_problem_1'

dados = []
for arquivo in os.listdir(diretorio):
    if arquivo.startswith('dataset_full_') and arquivo.endswith('.pkl'):
        tempo = float(arquivo.split('_')[2])  # aceita valores como "0.0", "16.6666", etc.
        with open(os.path.join(diretorio, arquivo), 'rb') as f:
            df = dill.load(f)
        temp = df[['fck', 'rh', 'cov', 'lambda 1', 'lambda 2', 'lambda 3', 'lambda 4']].copy()
        temp['tempo'] = tempo
        temp.columns = ['fck', 'rh', 'cov', 'lambda1', 'lambda2', 'lambda3', 'lambda4', 'tempo']
        dados.append(temp)
        print(f" {arquivo} - {len(temp)} registros")

df_final = pd.concat(dados, ignore_index=True).sort_values('tempo')
conn = sqlite3.connect('durability_data.db')
df_final.to_sql('durability_data', conn, if_exists='replace', index=False)
conn.close()
print(f"\n Banco criado com {len(df_final)} registros")
```
