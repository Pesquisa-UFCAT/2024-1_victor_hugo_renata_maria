```python
def co2_atmosferico_1900_1950(ano):
    """
    CO2 atmosférico (%) – período histórico inicial (1900–1950)
    """

    if ano < 1900 or ano > 1950:
        raise ValueError("Modelo válido apenas para 1900–1950")

    # crescimento médio ~0.30 ppm/ano
    co2_ppm = 296.0 + 0.30 * (ano - 1900)
    return co2_ppm / 1e4
```

```python
def co2_atmosferico_1950_2000(ano):
    """
    CO2 atmosférico (%) – período histórico industrial (1950–2000)
    """

    if ano < 1950 or ano > 2000:
        raise ValueError("Modelo válido apenas para 1950–2000")

    # crescimento médio ~1.16 ppm/ano
    co2_ppm = 311.0 + 1.16 * (ano - 1950)
    return co2_ppm / 1e4
```

```python
def co2_atmosferico_pos2000(ano):
    """
    CO2 atmosférico (%) – período moderno (>=2000)
    """

    if ano < 2000:
        raise ValueError("Modelo válido apenas para anos >= 2000")

    t = ano - 2000

    C0 = 369.0   # ppm em 2000
    a  = 1.85    # ppm/ano
    b  = 0.018   # ppm/ano²

    co2_ppm = C0 + a * t + b * t**2
    return co2_ppm / 1e4
```

```python
def co2_atmosferico_ano(ano):
    """
    Concentração média global de CO2 atmosférico (%)
    válida de 1900 em diante.

    Modelo por regimes históricos e modernos.
    """

    if ano < 1900:
        raise ValueError("Modelo válido apenas para anos >= 1900")

    if ano <= 1950:
        return co2_atmosferico_1900_1950(ano)
    elif ano <= 2000:
        return co2_atmosferico_1950_2000(ano)
    else:
        return co2_atmosferico_pos2000(ano)
```

```python
def co2_atmosferico_estrutura(t_estrutura, ano_inicio):
    """
    CO2 atmosférico (%) ao longo da vida da estrutura
    """

    ano_calendario = ano_inicio + t_estrutura
    return co2_atmosferico_ano(ano_calendario)
```

```python
anos_teste = [1900, 1950, 2000, 2010, 2020, 2023]

for ano in anos_teste:
    print(f"{ano}: {co2_atmosferico_ano(ano):.5f} %")
```
