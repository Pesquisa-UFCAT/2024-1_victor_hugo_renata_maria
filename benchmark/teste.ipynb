# Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Model

```python
tempo = np.linspace(0, 100, 100)
g = np.sin(tempo) + np.random.normal(0, 0.1, 100)
g
```

```python
plt.figure(figsize=(10, 6))
scatter = plt.scatter(tempo, g, c=tempo, cmap='viridis', s=50)
plt.colorbar(scatter, label='Tempo')
plt.xlabel('Tempo')
plt.ylabel('Valor do Array')
plt.title('Evolução Temporal do Array')
plt.grid(True, alpha=0.3)
plt.show()
```
