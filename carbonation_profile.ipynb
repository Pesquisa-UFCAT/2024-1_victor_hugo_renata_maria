# Profile for test

### Data

```python
CO2_ppm = 800               # Change as you wish. In ppm units        
fc = 30.0                   # Change as you wish. In MPa units 
rh = 35.4                   # Change as you wish. In % units
cement_type = 0             # Change as you wish. 0 to 5. CPII - Z=0, CPV - ARI=1, CPIV=2, CPII - F=3, CPIII=4, CPII - E=5
exposure_conditions = 0     # Change as you wish. 0 to 2. PIA=0, UEA=1, PEA=2
```

```python
df = {'t (years)': list(range(1, 101))}
df = pd.DataFrame(df)
df['CO2 (%)'] = CO2_ppm / 10000
df['fc (MPa)'] = fc
df['RH (%)'] = rh
df['Type of cement'] = cement_type
df['Exposure conditions'] = exposure_conditions
expected_order = list(model.feature_names_in_)
df = df[expected_order]
df
```

### Creating profile

```python
df_aux = df.copy()
y_pred = model.predict(df)
df_aux['carb. depth (mm)'] = list(y_pred)
df_aux
```
