```python
# Importa as bibliotecas necessárias
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import itertools
```

```python
# Lê a planilha com os dados
arquivo_excel = 'final_df_1.xlsx'
df = pd.read_excel(arquivo_excel)
```

```python
x = df.iloc[:, 7]  # 8ª coluna pelo índice (base zero)
y = df.iloc[:, 8]  # 9ª coluna pelo índice (base zero)

# Criar o gráfico de dispersão
plt.scatter(x, y)

# Configuração do gráfico
plt.xlabel('time')
plt.ylabel('pf')
plt.grid()

# Mostrar o gráfico
plt.show()
```

```python
# Remove a coluna 'm_rd' do DataFrame f permanentemente.
df.drop(columns='m_rd', inplace=True)

# Calcula a matriz de correlação para as colunas numéricas do DataFrame df.
df.corr()

# Aumenta o tamanho do mapa de calor para facilitar a visualização.
plt.figure(figsize=(16, 6))

# Armazena o objeto do mapa de calor em uma variável para facilitar modificações futuras, como adicionar mais características.
# Define o intervalo de valores exibidos no mapa de cores de -1 a 1 e exibe os valores das correlações no mapa.
heatmap = sns.heatmap(df.corr(), vmin=-1, vmax=1, annot=True)

# Adiciona um título ao mapa de calor. O parâmetro pad define a distância do título em relação ao topo do mapa.
heatmap.set_title('Mapa de Calor das Correlações', fontdict={'fontsize':12}, pad=12);
```

```python
# Aumenta o tamanho da figura para facilitar a visualização do mapa de calor.
plt.figure(figsize=(16, 6))

# Cria um mapa de calor baseado na matriz de correlação do DataFrame df.
# Define o intervalo de valores exibidos no mapa de cores de -1 a 1, exibe os valores das correlações no mapa
# e aplica a paleta de cores 'BrBG' para personalizar a aparência.
heatmap = sns.heatmap(df.corr(), vmin=-1, vmax=1, annot=True, cmap='BrBG')

# Adiciona um título ao mapa de calor com tamanho de fonte especificado.
# O parâmetro pad ajusta a distância entre o título e o topo do mapa.
heatmap.set_title('Mapa de Calor das Correlações', fontdict={'fontsize':18}, pad=12);

# Salva o mapa de calor como um arquivo .png.
# dpi - define a resolução da imagem salva em pontos por polegada (300 dpi neste caso).
# bbox_inches - quando definido como 'tight', evita que os rótulos sejam cortados na imagem salva.
plt.savefig('heatmap.png', dpi=300, bbox_inches='tight')
```

```python
# Calcula a matriz de correlação para as colunas numéricas do DataFrame df,
# e seleciona a coluna 'pf' para ordenar as correlações com ela em ordem decrescente.
df.corr()[['pf']].sort_values(by='pf', ascending=False)

# Aumenta o tamanho da figura para facilitar a visualização do mapa de calor.
plt.figure(figsize=(8, 12))

# Cria um mapa de calor baseado na correlação entre as features e a variável 'pf'.
# Ordena as correlações da coluna 'pf' em ordem decrescente, e define o intervalo de valores de -1 a 1.
# Exibe os valores das correlações no mapa e utiliza a paleta de cores 'BrBG' para personalizar a aparência.
heatmap = sns.heatmap(df.corr()[['pf']].sort_values(by='pf', ascending=False), vmin=-1, vmax=1, annot=True, cmap='BrBG')

# Adiciona um título ao mapa de calor, destacando as features correlacionadas com 'pf'.
# O parâmetro pad ajusta a distância entre o título e o topo do mapa.
heatmap.set_title('Features Correlacionadas com "pf"', fontdict={'fontsize':18}, pad=16);
```
