### State limit function t = 0

```python
n_latent_samples = 2000
n_samples = 100
r   = Normal(loc = 5., scale=0.8)
s   = Normal(loc = 2., scale=0.6)
joint = JointIndependent(marginals=[r, s])
x_val = joint.rvs(n_samples)
submatrices = np.array_split(x_val, 2)
```

```python
y, df = g_toy_problem_parallel_with_multiprocessing(submatrices, n_latent_samples, t=0)
```

```python
r = []
s = []
lambda_1 = []
lambda_2 = []
lambda_3 = []
lambda_4 = []

for i in range(len(df)):
    r.append(df[i]["r"].iloc[-1])
    s.append(df[i]["s"].iloc[-1])

    lambda_1.append(y[i, 0])
    lambda_2.append(y[i, 1])
    lambda_3.append(y[i, 2])
    lambda_4.append(y[i, 3])

lambdas = [lambda_1, lambda_2, lambda_3, lambda_4]

for i, lam in enumerate(lambdas, start=1):
    plt.figure(figsize=(8, 6))

    plt.scatter(
        r,
        s,
        c=lam,
        cmap='viridis',
        s=40
    )

    plt.xlabel('r')
    plt.ylabel('s')
    plt.colorbar(label=f'λ{i}')

    plt.tight_layout()
    plt.show()


```
