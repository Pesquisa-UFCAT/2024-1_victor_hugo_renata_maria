```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Gerar Dados ---
x = np.linspace(0, 4 * np.pi, 500) # De 0 a 4pi

# Funções
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.sin(x) + np.cos(x)

# --- 2. Configurar a Grade (4 linhas, 1 coluna) ---
# sharex=True: Faz todos compartilharem o eixo X (só o último mostra os números)
# figsize: (largura, altura) -> Mais alto para caber 4 gráficos
fig, axes = plt.subplots(nrows=4, ncols=1, sharex=True, figsize=(8, 5))

# --- 3. Plotagem ---

# Gráfico 1: Seno
axes[0].plot(x, y1, color='tab:blue', linewidth=2)
axes[0].set_ylabel(r'$\sin(x)$', fontsize=12)
# axes[0].set_title('Comparação de Funções Trigonométricas', fontsize=14) # Título só no topo
axes[0].grid(True, alpha=0.3)

# Gráfico 2: Cosseno
axes[1].plot(x, y2, color='tab:orange', linewidth=2)
axes[1].set_ylabel(r'$\cos(x)$', fontsize=12)
axes[1].grid(True, alpha=0.3)

# Gráfico 3: Tangente
axes[2].plot(x, y3, color='tab:green', linewidth=2)
axes[2].set_ylabel(r'$\tan(x)$', fontsize=12)
# IMPORTANTE: Tangente vai ao infinito, então cortamos o eixo Y para ficar bonito
axes[2].set_ylim(-3, 3) 
axes[2].grid(True, alpha=0.3)

# Gráfico 4: Seno + Cosseno
axes[3].plot(x, y4, color='tab:red', linewidth=2)
axes[3].set_ylabel(r'$\sin(x) + \cos(x)$', fontsize=12)
axes[3].set_xlabel('Eixo X (Radianos)', fontsize=12) # Label X só no último
axes[3].grid(True, alpha=0.3)

# --- 4. Ajustes Finais ---
plt.tight_layout() # Arruma os espaçamentos automaticamente
plt.show()
```

```python
def plot_gld_vs_data(lambdas, n_points=5000, quantile_trim=1e-3, qp_min=1e-8, ylim=None):
    l1, l2, l3, l4 = lambdas

    # Curva da PDF GLD (robusta para evitar explosões numéricas)
    a = float(quantile_trim)
    u = np.linspace(a, 1.0 - a, n_points)
    x = gld_fkml_quantile(u, l1, l2, l3, l4)
    qp = gld_fkml_qprime(u, l2, l3, l4)

    mask = np.isfinite(x) & np.isfinite(qp) & (qp > qp_min)
    x, f = x[mask], (1.0 / qp[mask])

    order = np.argsort(x)
    x, f = x[order], f[order]

    return list(x), list(f)

glam_data, _ = plot_gld_vs_data(test_lambda_values) 
```
