# Dataset from real real function

### General data

```python
n_latent_samples = 5000
n_samples = 1000
r   = Normal(loc = 5., scale=0.8)
s   = Normal(loc = 2., scale=0.6)
```

### State limit function $t$ = 0

```python
step_time = 0.0
divs = 10
```

```python
joint = JointIndependent(marginals=[r, s])
x_val = joint.rvs(n_samples)
submatrices = np.array_split(x_val, divs)
df, glam_data = g_toy_problem_parallel_with_multiprocessing(submatrices, n_latent_samples, t=step_time)
glam_data
```
